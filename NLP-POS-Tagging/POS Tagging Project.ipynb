```python
import nltk
from nltk import word_tokenize
from nltk import pos_tag
from nltk.tree import Tree

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
def process_language(text, language_name):
    print(f"\n--- {language_name} POS Tagging ---")

    # Tokenization
    tokens = text.split()   # simple & safe

    # POS tagging
    tags = pos_tag(tokens)

    # Print tags
    for word, tag in tags:
        print(f"{word} → {tag}")

    # Convert to tree format
    tagged_str = ""
    for word, tag in tags:
        tagged_str += f"({tag} {word}) "

    tagged_str = tagged_str.strip()

    # Display tree
    tree = Tree.fromstring(f"(S {tagged_str})")
    tree.pretty_print()
```

```python
while True:
    print("\nSelect Language:")
    print("1. Hindi")
    print("2. Bengali")
    print("3. Telugu")
    print("4. Tamil")
    print("5. Marathi")
    print("6. English")
    print("7. Exit")

    choice = input("Enter choice: ")

    if choice == "7":
        print("Exiting...")
        break

    sentence = input("Enter sentence: ")

    if choice == "1":
        process_language(sentence, "Hindi")

    elif choice == "2":
        process_language(sentence, "Bengali")

    elif choice == "3":
        process_language(sentence, "Telugu")

    elif choice == "4":
        process_language(sentence, "Tamil")

    elif choice == "5":
        process_language(sentence, "Marathi")

    elif choice == "6":
        process_language(sentence, "English")

    else:
        print("Invalid choice")
```
